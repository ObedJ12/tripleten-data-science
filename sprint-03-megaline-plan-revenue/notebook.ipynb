```python
def get_revenue(row):
    extra_minutes = max(row['minutes'] - row['monthly_minutes'], 0)
    extra_messages = max(row['messages'] - row['messages_included'], 0)

    included_gb = row.get('gb_included', 0)
    used_gb = np.ceil(row['gb_used'])
    extra_internet = max(used_gb - included_gb, 0)

    cost_extra_minutes = extra_minutes * row['usd_per_minute']
    cost_extra_messages = extra_messages * row['usd_per_message']
    cost_extra_internet = extra_internet * row['usd_per_gb']

    return row['usd_monthly_pay'] + cost_extra_minutes + cost_extra_messages + cost_extra_internet
```
