# Analyze patterns in missing values
The dataset shows distinct patterns in missing values across several columns. Minimal missing data is observed in the name and genre columns, each with approximately 1.2% missing values. These likely result from occasional data entry omissions and can be handled by dropping or imputing a small number of records. In contrast, the critic_score and user_score columns have over 50% missing values, while the rating column is missing around 40.6%. These high-missingness columns are typically associated with subjective or external sources—such as reviews and content ratings—which may not be available for all games, especially older titles, indie releases, or those from regions without standardized rating systems. Rather than discarding a large portion of the data, it would be more practical to create binary indicators for the presence of these values or consider imputation strategies if they are needed for modeling. Further analysis by platform or release year could reveal more about the distribution and causes of the missing data.

````python
# Handle missing values based on analysis
# Your code here to handle missing values according to your strategy

# Step 1: Drop rows where essential categorical data is missing
games = games.dropna(subset=['name', 'genre'])

# Step 2: Add missing value indicators for critical columns (optional but useful for ML models)
games['has_critic_score'] = games['critic_score'].notna().astype(int)
games['has_user_score'] = games['user_score'].notna().astype(int)
games['has_rating'] = games['rating'].notna().astype(int)
````

The missing values in the dataset likely stem from a combination of data availability, external dependencies, and historical practices. For instance, the high percentage of missing values in critic_score and user_score (over 50%) is probably because not all games are reviewed by critics or rated by users. This is especially true for older titles, lesser-known games, or those released in regions where such reviews were not consistently collected. Similarly, the rating column may be missing for games that were never rated by an official content rating board like the ESRB, which is more common outside North America or for digital-only or indie releases.
We chose to leave the missing values in critic_score, user_score, and rating as-is rather than fill them. This is because imputing values in such subjective fields could introduce bias or misleading information. Instead, we added binary indicator columns (has_critic_score, has_user_score, has_rating) to capture the presence or absence of these values. This allows any downstream analysis or modeling to account for the missingness as a potential feature. Conversely, we dropped rows with missing name or genre since these are essential for identifying and categorizing the games, and only a very small percentage of rows were affected.

## 4.1 Regional Platform Analysis

Let's begin by examining platform performance across different regions:

````python
# Function to analyze platform performance by region
def analyze_platform_by_region(df, region_column):
    """
    Analyzes total sales by platform for a given region.
    
    Parameters:
    - df: DataFrame with platform and regional sales data
    - region_column: str, column name for the region (e.g., 'na_sales')
    
    Returns:
    - DataFrame sorted by total sales for the specified region
    """
    region_sales = df.groupby('platform')[region_column].sum().sort_values(ascending=False)
    return region_sales.reset_index().rename(columns={region_column: 'total_sales'})
````

````python
# Analyze each region
regions = ['na_sales', 'eu_sales', 'jp_sales', 'other_sales']
regional_performance = {}

for region in regions:
    regional_performance[region] = analyze_platform_by_region(games, region)
````

# Step 5 : Hypothesis Tests

—Average user ratings of the Xbox One and PC platforms are the same.

## Hypothesis Testing: User Ratings

---

### 1. Average User Ratings: Xbox One vs. PC

**Hypotheses:**
- **Null Hypothesis (H₀):** The average user ratings for Xbox One and PC platforms are equal.
- **Alternative Hypothesis (H₁):** The average user ratings for Xbox One and PC platforms are different.

**Test Used:** Independent two-sample t-test (`scipy.stats.ttest_ind`)  
**Significance Level (α):** 0.05  
**p-value:** 0.123

**Conclusion:**  
Since the p-value (0.123) > α (0.05), we **fail to reject the null hypothesis**.  
There is **no statistically significant difference** in average user ratings between Xbox One and PC platforms.

---


````python
# Set alpha
alpha = 0.05

# Filter ratings for Xbox One and PC (exclude missing values)
xone_ratings = df_relevant[
    (df_relevant['platform'] == 'XOne') & (df_relevant['user_score'].notna())
]['user_score']

pc_ratings = df_relevant[
    (df_relevant['platform'] == 'PC') & (df_relevant['user_score'].notna())
]['user_score']

# Perform independent t-test
t_stat1, p_val1 = ttest_ind(xone_ratings, pc_ratings, equal_var=False)

# Print result
print("Xbox One vs PC - p-value:", p_val1)
if p_val1 < alpha:
    print("Reject H0: Ratings differ significantly.")
else:
    print("Fail to reject H0: No significant difference in ratings.")
````

### 2. Average User Ratings: Action vs. Sports Genres

**Hypotheses:**
- **Null Hypothesis (H₀):** The average user ratings for Action and Sports genres are equal.
- **Alternative Hypothesis (H₁):** The average user ratings for Action and Sports genres are different.

**Test Used:** Independent two-sample t-test (`scipy.stats.ttest_ind`)  
**Significance Level (α):** 0.05  
**p-value:** 0.00006

**Conclusion:**  
Since the p-value (0.00006) < α (0.05), we **reject the null hypothesis**.  
There **is a statistically significant difference** in average user ratings between the Action and Sports genres.

````python
# Filter ratings for Action and Sports genres (exclude missing values)
action_ratings = df_relevant[
    (df_relevant['genre'] == 'Action') & (df_relevant['user_score'].notna())
]['user_score']

sports_ratings = df_relevant[
    (df_relevant['genre'] == 'Sports') & (df_relevant['user_score'].notna())
]['user_score']

# Perform independent t-test
t_stat2, p_val2 = ttest_ind(action_ratings, sports_ratings, equal_var=False)

# Print result
print("Action vs Sports - p-value:", p_val2)
if p_val2 < alpha:
    print("Reject H0: Ratings differ significantly.")
else:
    print("Fail to reject H0: No significant difference in ratings.")
````
