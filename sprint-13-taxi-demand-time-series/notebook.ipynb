## Training

```python
def make_features(df, max_lag=24, rolling_windows=(3, 6, 24)):
    df = df.copy()

    # Calendar features
    df['hour'] = df.index.hour
    df['dayofweek'] = df.index.dayofweek
    df['day'] = df.index.day
    df['month'] = df.index.month

    # Lag features
    for lag in range(1, max_lag + 1):
        df[f'lag_{lag}'] = df['num_orders'].shift(lag)

    # Rolling means based only on past values
    for window in rolling_windows:
        df[f'rolling_mean_{window}'] = df['num_orders'].shift(1).rolling(window).mean()

    return df

featured = make_features(data_hourly, max_lag=24, rolling_windows=(3, 6, 24)).dropna()

display(featured.head())
print('Feature table shape:', featured.shape)
```

```python
# features and target
X = featured.drop('num_orders', axis=1)   # use 'num_orders' as your target
y = featured['num_orders']

# first split: train+val and test (10%)
X_temp, X_test, y_temp, y_test = train_test_split(
    X, y, test_size=0.1, random_state=42
)

# second split: train and validation
X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp, test_size=0.2, random_state=42
)

print(X_train.shape, X_val.shape, X_test.shape)
```

```python
# Linear Regression
lr = LinearRegression()
lr.fit(X_train, y_train)

val_pred_lr = lr.predict(X_val)
lr_rmse = np.sqrt(mean_squared_error(y_val, val_pred_lr))

# Random Forest (baseline)
rf = RandomForestRegressor(random_state=42)
rf.fit(X_train, y_train)

val_pred_rf = rf.predict(X_val)
rf_rmse = np.sqrt(mean_squared_error(y_val, val_pred_rf))

print("LR validation RMSE:", lr_rmse)
print("RF validation RMSE:", rf_rmse)
```

```python
best_rmse = float('inf')
best_model = None

for n in [50, 100, 200]:
    for depth in [5, 10, None]:
        model = RandomForestRegressor(
            n_estimators=n,
            max_depth=depth,
            random_state=42
        )
        model.fit(X_train, y_train)
        
        preds = model.predict(X_val)
        rmse = np.sqrt(mean_squared_error(y_val, preds))
        
        if rmse < best_rmse:
            best_rmse = rmse
            best_model = model

print("Best validation RMSE:", best_rmse)
```

## Testing

```python
if best_rmse < lr_rmse:
    final_model = best_model
    print("Selected model: Random Forest")
else:
    final_model = lr
    print("Selected model: Linear Regression")
```

```python
# combine datasets
X_full = pd.concat([X_train, X_val])
y_full = pd.concat([y_train, y_val])

final_model.fit(X_full, y_full)
```

```python
test_pred = final_model.predict(X_test)
test_rmse = np.sqrt(mean_squared_error(y_test, test_pred))

print("Final TEST RMSE:", test_rmse)
print("Project requirement: RMSE ≤ 48")
print("Requirement met:", "✅ Yes" if test_rmse <= 48 else "❌ No")
```
